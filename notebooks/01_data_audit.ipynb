```python
tables = [
    "bureau.csv",
    "bureau_balance.csv",
    "previous_application.csv",
    "POS_CASH_balance.csv",
    "installments_payments.csv",
    "credit_card_balance.csv"
]

for table in tables:
    df_sample = pd.read_csv(
        DATA_DIR / table,
        nrows=5
    )

    print("\n", "=" * 60)
    print(table)
    print("Columns:", len(df_sample.columns))
    print(df_sample.columns.tolist())
```

```python
def count_rows(filepath):
    with open(filepath, "rb") as f:
        return sum(1 for _ in f) - 1

for table in tables:
    path = DATA_DIR / table
    print(table, count_rows(path))
```

```python
# ============================================
# DAY 1 — RELATIONSHIP / DATA ARCHITECTURE
# ============================================

# 1. Check applicant IDs
print("TRAIN applicants:", train["SK_ID_CURR"].nunique())
print("TEST applicants :", test["SK_ID_CURR"].nunique())

# 2. Check overlap between train and test
train_ids = set(train["SK_ID_CURR"])
test_ids = set(test["SK_ID_CURR"])

print("Train/Test ID overlap:", len(train_ids & test_ids))

# 3. Check historical tables' applicant coverage
for table in [
    "bureau.csv",
    "previous_application.csv",
    "POS_CASH_balance.csv",
    "installments_payments.csv",
    "credit_card_balance.csv"
]:
    df_sample = pd.read_csv(DATA_DIR / table, nrows=100000)

    coverage = df_sample["SK_ID_CURR"].nunique()

    print(f"{table:30s} unique applicants in sample: {coverage:,}")

# 4. Check bureau -> bureau_balance relationship
bureau_sample = pd.read_csv(
    DATA_DIR / "bureau.csv",
    nrows=100000
)

bureau_balance_sample = pd.read_csv(
    DATA_DIR / "bureau_balance.csv",
    nrows=100000
)

print("\nBureau IDs in sample:",
      bureau_sample["SK_ID_BUREAU"].nunique())

print("Bureau balance IDs in sample:",
      bureau_balance_sample["SK_ID_BUREAU"].nunique())

print(
    "Bureau IDs appearing in bureau_balance sample:",
    len(
        set(bureau_sample["SK_ID_BUREAU"])
        &
        set(bureau_balance_sample["SK_ID_BUREAU"])
    )
)

# 5. Check previous application -> installment/POS/credit card linkage
previous_sample = pd.read_csv(
    DATA_DIR / "previous_application.csv",
    nrows=100000
)

print("\nPrevious application IDs:",
      previous_sample["SK_ID_PREV"].nunique())

for table in [
    "POS_CASH_balance.csv",
    "installments_payments.csv",
    "credit_card_balance.csv"
]:
    df_sample = pd.read_csv(DATA_DIR / table, nrows=100000)

    overlap = len(
        set(previous_sample["SK_ID_PREV"])
        &
        set(df_sample["SK_ID_PREV"])
    )

    print(
        f"{table:30s} SK_ID_PREV overlap: {overlap:,}"
    )
```
